```python
# ==========================================
#  CREATE CLEAN DATASET
# ==========================================

df_clean = df.copy()

print("Nombre de lignes :", len(df_clean))
print("Nombre de colonnes :", len(df_clean.columns))
```

```python
# ==========================================
#  FEATURE ENGINEERING - FINANCIAL METRICS
# ==========================================

# Revenue generated by sold units
df_clean["Revenue"] = (
    df_clean["Units_Sold"] * df_clean["Unit_Price"]
)

# Cost of sold units
df_clean["Cost_of_Goods_Sold"] = (
    df_clean["Units_Sold"] * df_clean["Unit_Cost"]
)

# Profit generated by sold units
df_clean["Profit"] = (
    df_clean["Revenue"] - df_clean["Cost_of_Goods_Sold"]
)

# Profit margin
df_clean["Profit_Margin"] = (
    df_clean["Profit"] / df_clean["Revenue"]
)

# Inventory value
df_clean["Inventory_Value"] = (
    df_clean["Inventory_Level"] * df_clean["Unit_Cost"]
)

print(df_clean[
    [
        "Units_Sold",
        "Unit_Cost",
        "Unit_Price",
        "Revenue",
        "Cost_of_Goods_Sold",
        "Profit",
        "Profit_Margin",
        "Inventory_Value"
    ]
].head())
```

```python
# ==========================================
# FEATURE ENGINEERING - SUPPLY CHAIN METRICS
# ==========================================

# Difference between actual demand and forecasted demand
df_clean["Demand_Gap"] = (
    df_clean["Units_Sold"] - df_clean["Demand_Forecast"]
)

# Absolute forecast error
df_clean["Forecast_Error"] = (
    abs(df_clean["Units_Sold"] - df_clean["Demand_Forecast"])
)

# Inventory coverage in days
df_clean["Inventory_Coverage_Days"] = (
    df_clean["Inventory_Level"] /
    df_clean["Demand_Forecast"].replace(0, np.nan)
)

# Difference between current inventory and reorder point
df_clean["Reorder_Gap"] = (
    df_clean["Inventory_Level"] - df_clean["Reorder_Point"]
)

# Potential inventory risk
df_clean["Inventory_Risk"] = np.where(
    df_clean["Inventory_Level"] < df_clean["Reorder_Point"],
    "Below Reorder Point",
    "Normal"
)

print(
    df_clean[
        [
            "Units_Sold",
            "Demand_Forecast",
            "Demand_Gap",
            "Forecast_Error",
            "Inventory_Level",
            "Reorder_Point",
            "Reorder_Gap",
            "Inventory_Coverage_Days",
            "Inventory_Risk"
        ]
    ].head(10)
)
```

```python
# ==========================================
# FEATURE ENGINEERING QUALITY CHECK
# ==========================================

print("===== 1. Valeurs manquantes =====")
print(df_clean[
    [
        "Revenue",
        "Cost_of_Goods_Sold",
        "Profit",
        "Profit_Margin",
        "Inventory_Value",
        "Demand_Gap",
        "Forecast_Error",
        "Inventory_Coverage_Days",
        "Reorder_Gap",
        "Inventory_Risk"
    ]
].isna().sum())


print("\n===== 2. Statistiques des nouvelles variables =====")
print(df_clean[
    [
        "Revenue",
        "Cost_of_Goods_Sold",
        "Profit",
        "Profit_Margin",
        "Inventory_Value",
        "Demand_Gap",
        "Forecast_Error",
        "Inventory_Coverage_Days",
        "Reorder_Gap"
    ]
].describe().T)


print("\n===== 3. Répartition du risque d'inventaire =====")
print(df_clean["Inventory_Risk"].value_counts())


print("\n===== 4. Vérification du profit négatif =====")
print(
    "Nombre de lignes avec Profit < 0 :",
    (df_clean["Profit"] < 0).sum()
)
```

```python
# ==========================================
# HANDLE UNDEFINED METRICS
# ==========================================

# Profit margin is undefined when there are no sales
df_clean["Profit_Margin"] = np.where(
    df_clean["Revenue"] > 0,
    df_clean["Profit"] / df_clean["Revenue"],
    np.nan
)

# Inventory coverage is undefined when forecasted demand is zero
df_clean["Inventory_Coverage_Days"] = np.where(
    df_clean["Demand_Forecast"] > 0,
    df_clean["Inventory_Level"] / df_clean["Demand_Forecast"],
    np.nan
)

print("Valeurs manquantes après correction :")

print(
    df_clean[
        [
            "Profit_Margin",
            "Inventory_Coverage_Days"
        ]
    ].isna().sum()
)
```

```python
print("===== ZERO DEMAND FORECAST =====")
print(
    "Demand_Forecast = 0 :",
    (df_clean["Demand_Forecast"] == 0).sum()
)

print("\n===== ZERO SALES =====")
print(
    "Units_Sold = 0 :",
    (df_clean["Units_Sold"] == 0).sum()
)

print("\n===== ZERO REVENUE =====")
print(
    "Revenue = 0 :",
    (df_clean["Revenue"] == 0).sum()
)
```

```python
# ==========================================
# DEMAND PERFORMANCE
# ==========================================

df_clean["Demand_Performance"] = np.select(
    [
        df_clean["Demand_Gap"] > 0.10 * df_clean["Demand_Forecast"],
        df_clean["Demand_Gap"] < -0.10 * df_clean["Demand_Forecast"]
    ],
    [
        "Above Forecast",
        "Below Forecast"
    ],
    default="Close to Forecast"
)

print(df_clean["Demand_Performance"].value_counts())
```

```python
print(
    df_clean[
        [
            "Units_Sold",
            "Demand_Forecast",
            "Demand_Gap",
            "Demand_Performance"
        ]
    ].head(15)
)
```

```python
# ==========================================
# STOCK STATUS
# ==========================================

df_clean["Stock_Status"] = np.where(
    df_clean["Inventory_Level"] < df_clean["Reorder_Point"],
    "Below Reorder Point",
    "At/Above Reorder Point"
)

print("===== STOCK STATUS =====")
print(df_clean["Stock_Status"].value_counts())


print("\n===== VERIFICATION =====")
print(
    pd.crosstab(
        df_clean["Stock_Status"],
        df_clean["Inventory_Risk"]
    )
)
```

```python
# ==========================================
# FINAL DATASET CHECK
# ==========================================

print("Shape finale :", df_clean.shape)

print("\nColonnes finales :")
print(df_clean.columns.tolist())

print("\nValeurs manquantes :")
print(df_clean.isna().sum())

print("\nTypes de données :")
print(df_clean.dtypes)
```

```python
# ==========================================
# CORRECTION DE DEMAND PERFORMANCE
# ==========================================

df_clean["Demand_Performance"] = np.select(
    [
        df_clean["Demand_Forecast"] == 0,
        df_clean["Demand_Gap"] > 0.10 * df_clean["Demand_Forecast"],
        df_clean["Demand_Gap"] < -0.10 * df_clean["Demand_Forecast"]
    ],
    [
        "No Forecast",
        "Above Forecast",
        "Below Forecast"
    ],
    default="Close to Forecast"
)

print(df_clean["Demand_Performance"].value_counts())
```

```python
# ==========================================
# EDA 15 - ANALYSE DES COMMANDES
# ==========================================

order_analysis = (
    df_clean
    .groupby("Stock_Status")
    .agg(
        Total_Orders=("Order_Quantity", lambda x: (x > 0).sum()),
        Total_Order_Quantity=("Order_Quantity", "sum"),
        Average_Order_Quantity=("Order_Quantity", lambda x: x[x > 0].mean()),
        Average_Inventory=("Inventory_Level", "mean"),
        Average_Reorder_Point=("Reorder_Point", "mean")
    )
    .reset_index()
)

print("Analyse des commandes selon le statut du stock :")
print(order_analysis)
```

```python
# ==========================================
# EDA 16 - LEAD TIME VS RISQUE DE STOCK
# ==========================================

lead_time_analysis = (
    df_clean
    .groupby("Supplier_ID")
    .agg(
        Average_Lead_Time=("Supplier_Lead_Time_Days", "mean"),
        Low_Stock_Observations=(
            "Stock_Status",
            lambda x: (x == "Below Reorder Point").sum()
        ),
        Low_Stock_Rate=(
            "Stock_Status",
            lambda x: (x == "Below Reorder Point").mean() * 100
        ),
        Average_Inventory=("Inventory_Level", "mean"),
        Total_Sales=("Units_Sold", "sum")
    )
    .reset_index()
    .sort_values("Average_Lead_Time", ascending=False)
)

print("Lead time et risque de stock par fournisseur :")
print(lead_time_analysis)
```

```python
# ==========================================
# EDA 17 - EVOLUTION DE LA DEMANDE DANS LE TEMPS
# ==========================================

monthly_demand = (
    df_clean
    .groupby(df_clean["Date"].dt.to_period("M"))
    .agg(
        Total_Sales=("Units_Sold", "sum"),
        Average_Daily_Sales=("Units_Sold", "mean"),
        Total_Profit=("Profit", "sum")
    )
    .reset_index()
)

monthly_demand["Date"] = monthly_demand["Date"].astype(str)

print("Évolution mensuelle de la demande :")
print(monthly_demand)
```

```python
# ==========================================
# EDA 18 - ANALYSE FINALE DES SKU
# ==========================================

final_sku_analysis = (
    df_clean
    .groupby("SKU_ID")
    .agg(
        Total_Sales=("Units_Sold", "sum"),
        Total_Profit=("Profit", "sum"),
        Average_Margin=("Profit_Margin", "mean"),
        Average_Inventory=("Inventory_Level", "mean"),
        Average_Coverage_Days=("Inventory_Coverage_Days", "mean"),
        Low_Stock_Rate=(
            "Stock_Status",
            lambda x: (x == "Below Reorder Point").mean() * 100
        ),
        Average_Forecast_Error=("Forecast_Error", "mean")
    )
    .reset_index()
)

print("Analyse finale des SKU :")
print(final_sku_analysis.head(10))
```

```python
print("\nTop 10 SKU par profit :")
print(
    final_sku_analysis
    .sort_values("Total_Profit", ascending=False)
    [["SKU_ID", "Total_Sales", "Total_Profit", "Average_Margin",
      "Average_Inventory", "Average_Coverage_Days",
      "Low_Stock_Rate", "Average_Forecast_Error"]]
    .head(10)
)
```
